```python
#1
def count_w():

    with open("documents.txt","r") as file:
        text=file.read()
    words=text.split()
    return len(words)

print(count_w())
```

```python
#2
def replace(old_word,new_word):
    with open("data.txt","r") as file:
        text=file.read()
    text=text.replace(old_word,new_word)
    with open("data.txt","w") as file:
        file.write(text)
replace("Teahouse","Mudhouse")
    
```

```python
#3
def reverse():
    with open("reverse.txt","r") as file:
        lines=file.readlines()
    for line in lines:
        print(line.strip()[::-1])
reverse()
```

```python
#4
def file_s():
    with open("stats.txt","r") as file:
        text=file.read()
    lines=text.splitlines()
    words=text.split()
    character=len(text)
    return len(lines),len(words),character
print(file_s())
```
