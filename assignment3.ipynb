```python
#1
import pandas as pd
import numpy as np
df=pd.DataFrame(np.random.randint(1,11,(5,3)),columns=["A","B","C"])
df["Product"]=df["A"]*df["B"]
print(df)
```

```python
#2
df=pd.DataFrame(np.random.randint(1,11,(4,3)),columns=["A","B","C"])
print(df)
print("Row wise sum:")
print(df.sum(axis=1))
print("Column wise sum:")
print(df.sum(axis=0))
```

```python
#3
df=pd.DataFrame(np.random.randint(1,11,(4,3)),columns=["l","m","n"])
df.iloc[1,0]=np.nan
df.iloc[3,1]=np.nan
df.iloc[2,2]=np.nan
print("Before")
print(df)
df=df.fillna(df.mean())
print("After:")
print(df)
```

```python
#4
df=pd.DataFrame(np.random.randint(1,11,(6,4)),columns=["l","m","n","o"])
df.iloc[1,0]=np.nan
df.iloc[4,2]=np.nan
print("Before")
print(df)
df=df.dropna
print("After")
print(df)
```

```python
#5
df=pd.DataFrame({"Category":np.random.choice(["A","B","C"],10),"Value":np.random.randint(10,101,10)})
print(df)
result=df.groupby("Category")["Value"].agg(["sum","mean"])
print(result)
```

```python
#6
df = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag", "Pencil", "Bottle"],
    "Category": np.random.choice(["A", "B", "C"], 5),
    "Sales": np.random.randint(100, 1000, 5)
})

print(df)

result = df.groupby("Category")["Sales"].sum()

print("Total sales:")
print(result)
```

```python
#7
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 85]
})

result = pd.merge(df1, df2, on="ID")

print(result)
```

```python
#8
index = [
    ["A", "A", "B", "B"],
    [1, 2, 1, 2]
]

multi_index = pd.MultiIndex.from_arrays(
    index,
    names=["Category", "Number"]
)

df = pd.DataFrame(
    {"Value": [10, 20, 30, 40]},
    index=multi_index
)

print(df)

print("Category A:")
print(df.loc["A"])

print("Category B:")
print(df.loc["B"])

print("First row:")
print(df.iloc[0])
```

```python
#9
categories = ["A", "A", "B", "B"]
subcategories = ["X", "Y", "X", "Y"]

index = pd.MultiIndex.from_arrays(
    [categories, subcategories],
    names=["Category", "SubCategory"]
)

df = pd.DataFrame(
    {"Value": np.random.randint(10, 101, 4)},
    index=index
)

print(df)

print("Sum:")
print(df.groupby(level=["Category", "SubCategory"]).sum())
```

```python
#10
df = pd.DataFrame(
    np.random.randint(1, 11, (5, 3)),
    columns=["A", "B", "C"]
)

print("Original:")
print(df)

result = df.apply(lambda x: x * 2)

print("After doubling:")
print(result)
```

```python
#11
df = pd.DataFrame(
    np.random.randint(1, 11, (6, 3)),
    columns=["A", "B", "C"]
)

df["Sum"] = df.apply(
    lambda row: row["A"] + row["B"] + row["C"],
    axis=1
)

print(df)
```

```python
#12
s = pd.Series([
    "apple",
    "banana",
    "mango",
    "orange",
    "grapes"
])

print("Original:")
print(s)

print("Uppercase:")
print(s.str.upper())
```
