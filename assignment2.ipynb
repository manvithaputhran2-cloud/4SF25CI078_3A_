```python
#1
import numpy as np
a=np.random.randint(1,21,(5,5))
print("Original Array:")
print(a)
a[:,2]=1
print("Replaced:")
print(a)
```

```python
#2
ar=np.arange(1,17).reshape(4,4)
print("original array:")
print(ar)
np.fill_diagonal(ar,0)
print("After replacing:")
print(ar)
```

```python
#3
ar=np.random.randint(1,21,(5,5))
print("Array:")
print(ar)
border=np.concatenate([ar[0,:],ar[-1,:],ar[1:-1,0],ar[1:-1,-1]])
print(border)
```

```python
#4
arr=np.arange(1,17).reshape(4,4)
print(arr)
row_sum=np.sum(arr,axis=1)
column_sum=np.sum(arr,axis=0)
print("Row sum")
print(row_sum)
print("Column sum")
print(column_sum)
```

```python
#5
arr=np.arange(1,10).reshape(3,3)
print("Original")
print(arr)
arr1=arr.reshape(1,9)
arr2=arr.reshape(9,1)
print("Shape (1,9):")
print(arr1)
print("Shape(9,1)")
print(arr2)
```

```python
#6
data_type=[("name","U20"),("age","i4"),("weight","f4")]
data=np.array([("Madhura",19,52.0),("Vibha",20,50.0),("Thanusha",20,45.0),("Sushma",19,45.5)],dtype=data_type)
print(data)
sorted=np.sort(data,order="age")
print(sorted)
```
